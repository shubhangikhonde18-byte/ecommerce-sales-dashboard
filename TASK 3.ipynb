```python
fruits = ["Apple","Banana","Mango"]

fruits.append("Orange")
print(fruits)

fruits.remove("Banana")
print(fruits)

fruits[0] = "Grapes"
print(fruits)

fruits.sort()
print(fruits)
print("Total fruits=",len(fruits))
```
