#OPEN FILE .txt file 

```python
file = open("student.txt","w")

file.write("Name: Shubhangi\n")
file.write("Course: MCA\n")
file.write("Skill: Python\n")

file.close()
```

##Reading a .txt File

```python
file = open("student.txt","r")

data = file.read()
print(data)
file.close()
```

#with Open file

```python
with open("student.txt","r") as file:
    data = file.read()
print(data)
    
```

#Creating the CSV File 

```python
import csv
```

```python
import csv

with open("student.csv","w",newline="") as file:
    writer = csv.writer(file)
    
    writer.writerow(["Name","Course","Marks"])
    writer.writerow(["Shubhangi","MCA",85])
    writer.writerow(["Rahul","BCA",78])
    writer.writerow(["Priya","MCA",90])
    writer.writerow(["Mina","MBA",86])
```

#CSV File Read

```python
with open("student.csv","r") as file:
  reader = csv.reader(file)
  for row in reader:
    print(row)
```

##CSV Data as Dictionary

```python
with open("student.csv","r") as file:
    reader = csv.DictReader(file)
    for row in reader:
        print(row)
        print(row["Name"])
```

##CSV मधून Specific Column Read करणे

```python
with open("student.csv","r") as file:
    reader = csv.DictReader(file)
    
    for row in reader:
        print(row["Name"])
```

#MCA Students Filter

```python
with open("student.csv","r") as file:
    reader = csv.DictReader(file)
    
    for row in reader:
        if row["Course"] == "MCA":
            print(row["Name"])
```
