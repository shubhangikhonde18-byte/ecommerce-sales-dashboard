##Create a NumPy Array

```python
import numpy as np
numbers = np.array([10,20,30,40,50])
print(numbers)
print(type(numbers))
```

##1D Array

```python
arr1 = np.array([10,20,30,40,50])
print(arr1)
```

##2D Array

```python
arr2 = np.array([[10, 20, 30],
                 [40, 50, 60]])

print(arr2)
print(arr2.ndim)
```

##Step 5 — NumPy Array Indexing

##Step 5 — NumPy Array Indexing

```python
arr1 = np.array([10,20,30,40,50])
print(arr1[-1])
```

##2D Array Indexing

```python
arr2 = np.array([[10, 20, 30],
                 [40, 50, 60]])
print(arr2[0,1])
print(arr2[1,1])
```

##Step 6 — NumPy Array Slicing

```python
arr1 = np.array([10,20,30,40,50])
print(arr1[1:4])
print(arr1[:3])
print(arr1[2:])
print(arr1[::2])
```

##2D Array Slicing

```python
arr2 = np.array([[10, 20, 30],
                 [40, 50, 60]])
print(arr2[0, :])
print(arr2[:, 0])
```

##Step 7 — NumPy Array Operations

```python
arr1 = np.array([10,20,30,40,50])
result = arr1 + 5
print(result)
```

##Multiplication

```python
result = arr1 * 2
print(result)
```

##Subtraction

```python
result = arr1 -5 
print(result)
```

##Division

```python
result = arr1 /10
print(result)
```

##Step 8 — दोन NumPy Arrays ची Operation

```python
a = np.array([10,20,30])
b = np.array([1,2,3])

print(a + b)
print(a - b)
print(a * b)
```
