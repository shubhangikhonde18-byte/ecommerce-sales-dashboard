##Step 1: Setup and Created TWO Datasets

```python
import pandas as pd

customers = pd.DataFrame({
    'customer_id': [1, 2, 3, 4, 5],
    'name': ['Amit', 'Priya', 'Rahul', 'Sneha', 'Rohan'],
    'city': ['Nagpur', 'Pune', 'Mumbai', 'Nagpur', 'Nashik']
})

orders = pd.DataFrame({
    'order_id': [101, 102, 103, 104, 105, 106],
    'customer_id': [1, 2, 2, 6, 3, 1],
    'amount': [500, 1200, 300, 800, 950, 400]
})

print(customers)
print(orders)
```

##Step 2: merge() ( SAME as SQL JOIN )

```python
# Inner: fakt matching rows
inner = pd.merge(customers, orders, on='customer_id', how='inner')
print(inner, sep='\n\n')
```

```python
# Left: sagle customers, order nasel tar NaN
left = pd.merge(customers, orders, on='customer_id', how='left')
print(left, sep='\n\n')
```

```python
# Right: sagle orders
right = pd.merge(customers, orders, on='customer_id', how='right')
print(right, sep='\n\n')
```

```python
# Outer: donhi kadche sagle rows
outer = pd.merge(customers, orders, on='customer_id', how='outer', indicator=True)

print(outer, sep='\n\n')
```

##If Different Column Name 

```python
pd.merge(customers, orders, left_on='customer_id', right_on='customer_id')
```

##Step 3: join() (index based)

```python
cust_idx = customers.set_index('customer_id')
ord_idx = orders.set_index('customer_id')

joined = cust_idx.join(ord_idx, how='left')
print(joined)
```

```python
cust_idx.join(ord_idx, how='inner', lsuffix='_cust', rsuffix='_ord')
```

##Step 4: concat() (stack)
##a Vertical (rows khali add)

```python
new_customers = pd.DataFrame({
    'customer_id': [6, 7],
    'name': ['Neha', 'Karan'],
    'city': ['Delhi', 'Nagpur']
})

all_customers = pd.concat([customers, new_customers], ignore_index=True)
print(all_customers)
```

##b Horizontal (columns shejari)

```python
side_by_side = pd.concat([customers, orders], axis=1)
print(side_by_side)
```

##(c) keys vapra (konta data konkadcha kalayla)

```python
pd.concat([customers, new_customers], keys=['old', 'new'])
```

##Step 6: Extra practice

```python
# Merge nantar analysis
merged = pd.merge(customers, orders, on='customer_id', how='inner')

# Total amount per city
print(merged.groupby('city')['amount'].sum())

# Konte customers ni kahich order kele nahit?
no_orders = pd.merge(customers, orders, on='customer_id', how='left', indicator=True)
print(no_orders[no_orders['_merge'] == 'left_only'])
```
