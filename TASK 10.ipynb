```python
import pandas as pd
```

##Create DataSet 

```python
data = {
    "Name":       ["Amit", "Sneha", "Rahul", "Priya", "Karan", "Neha", "Rohan", "Anjali"],
    "Department": ["Sales", "HR", "Sales", "IT", "IT", "HR", "IT", "Sales"],
    "City":       ["Nagpur", "Pune", "Nagpur", "Mumbai", "Pune", "Nagpur", "Mumbai", "Pune"],
    "Salary":     [30000, 45000, 28000, 60000, 55000, 42000, 65000, 33000],
    "Age":        [25, 32, 28, 35, 30, 29, 40, 27]
}

df = pd.DataFrame(data)
print(df)
```

```python
print(df.head())
print(df.shape)
print(df.info())
print(df.describe())
```

##Step 3: Rows filter

```python
# Salary 40000 peksha jast
high_salary = df[df["Salary"] > 40000]
print(high_salary)
```

```python
# Only IT department
it_dept = df[df["Department"] == "IT"]
print(it_dept)
```

```python
# Multiple conditions: AND (&) AND OR (|)
result = df[(df["Department"] == "IT") & (df["Salary"] > 55000)]
print(result)
```

```python
# Nagpur kiwa Pune madhle
cities = df[df["City"].isin(["Nagpur", "Pune"])]
print(cities)
```

```python
# query() vaprun pan karta yeta
result2 = df.query("Age > 28 and Salary < 50000")
print(result2)
```

##Step 4: Sorting kara

```python
# Salary ascending
df.sort_values("Salary")
print(df.sort_values("Salary"))
```

```python
# Descending 
df.sort_values("Salary", ascending=False)
print("\nSalary descending:")
print(df.sort_values("Salary", ascending=False))
```

```python
# Multiple columns: First Department, Then Salary descending
df.sort_values(["Department", "Salary"], ascending=[True, False])
print("\nDepartment (A-Z), mag Salary (jast pasun kami):")
print(df.sort_values(["Department", "Salary"], ascending=[True, False]))
```

```python

# Filter + sort ekatra
top_it = df[df["Department"] == "IT"].sort_values("Salary", ascending=False)
print("\nIT department, Salary descending:")
top_it = df[df["Department"] == "IT"].sort_values("Salary", ascending=False)
print(top_it)
```

##Step 5: groupby() and agg()

```python
# Simple: department-wise average salary
df.groupby("Department")["Salary"].mean()

# Multiple aggregations ekatra
summary = df.groupby("Department").agg(
    total_employees=("Name", "count"),
    avg_salary=("Salary", "mean"),
    max_salary=("Salary", "max"),
    min_salary=("Salary", "min"),
    total_salary=("Salary", "sum")
)
print(summary)
```

##Don columns var group karaycha asel:

```python
df.groupby(["Department", "City"]).agg(
    avg_salary=("Salary", "mean"),
    count=("Name", "count")
)
```

##Step 6: Sagla combine kara (final summary)

```python
# 1. Filter: fakt 25+ age che employees
filtered = df[df["Age"] > 25]

# 2. Group + aggregate
summary = filtered.groupby("Department").agg(
    employees=("Name", "count"),
    avg_salary=("Salary", "mean"),
    max_age=("Age", "max")
).reset_index()

# 3. Sort: avg salary jast pasun kami
summary = summary.sort_values("avg_salary", ascending=False)

print(summary)
```

##Step 7: Saving The Result

```python
summary.to_csv("summary.csv",index=False)
```
