##Step 1: Libraries install and import 

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid")   # sundar default look
```

##Step 2: Data load

```python
import pandas as pd

df = pd.read_excel(r"C:\Users\khond\OneDrive\Desktop\Shubhu Excel sheets\marksheet 1.xlsx", header=None)

for i in range(len(df)):
    if "Java" in df.iloc[i].astype(str).str.strip().tolist():
        header_row = i
        break

df.columns = df.iloc[header_row].astype(str).str.strip()
df = df.iloc[header_row + 1:].reset_index(drop=True)
df = df.dropna(how="all")
df = df.dropna(axis=1, how="all")

print(df.columns.tolist())
print(df.head())
```

##Step 1: Marks numbers madhe convert kara

```python
subjects = ["Java", "SEPM", "ADBMS", "CHN", "NS"]
df[subjects] = df[subjects].apply(pd.to_numeric, errors="coerce") ##marks covert into numbers

print(df[subjects].describe())
```

##Step 2: Seaborn import kara

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid")
```

##Step 3: Distribution Plot (marks kase pasarlet)

```python
fig, axes = plt.subplots(2, 3, figsize=(14, 8))
for ax, sub in zip(axes.flatten(), subjects):
    sns.histplot(df[sub], kde=True, bins=5, ax=ax)
    ax.set_title(sub)
axes.flatten()[-1].axis("off")
plt.tight_layout()
plt.savefig("distribution.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
##Step 4: Boxplot (subjects chi tulna)
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df[subjects])
plt.title("Marks by Subject")
plt.ylabel("Marks")
plt.savefig("boxplot.png", dpi=300, bbox_inches="tight")
plt.show()
```

##Step 5: Heatmap (subjects madhe sambandh)

```python
corr = df[subjects].corr()

plt.figure(figsize=(7, 5))
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5)
plt.title("Correlation Heatmap of Subjects")
plt.savefig("heatmap.png", dpi=300, bbox_inches="tight")
plt.show()
```
