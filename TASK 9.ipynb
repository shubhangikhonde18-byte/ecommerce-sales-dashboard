##Step 1: Create Messy sample dataset 

```python
import pandas as pd
import numpy as np

data = {
    "Name":   ["Amit", "Sneha", "Rahul", "Sneha", "Priya", "amit ", None, "Rahul"],
    "Age":    [25, 28, np.nan, 28, 22, 25, 30, np.nan],
    "City":   ["Pune", "Mumbai", "Nagpur", "Mumbai", None, "Pune", "Nashik", "Nagpur"],
    "Salary": [50000, 62000, 45000, 62000, np.nan, 50000, 58000, 45000],
}
df = pd.DataFrame(data)
print(df)
```

##Step 2: Missing values shodha (isnull)

```python
print(df.isnull())  
print(df.isnull().sum())
print(df.info())
```

##Step 3: Missing values handle fillna()

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())
df["City"] = df["City"].fillna("Unknown")
```

##dropna(): missing rows kadhun taka:

```python
df_clean = df.dropna()
df_clean = df.dropna(subset=["Name"])
```

##Step 4: Duplicates kadha

```python
df["Name"] = df["Name"].str.strip().str.title()

print(df.duplicated().sum())
print(df[df.duplicated(keep=False)])

df = df.drop_duplicates()
df = df.drop_duplicates(subset=["Name"],keep="first")
```

##Final check

```python
print(df.isnull().sum())
print(df.duplicated().sum())
print(df.shape)
```
